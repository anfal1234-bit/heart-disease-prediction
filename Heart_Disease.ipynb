### Task 2: Heart Disease System
**Dataset:** UCI Heart Disease (Cleveland)
**Selected Attributes:** Age, Sex, Chest Pain Type (cp), Max Heart Rate (thalach)
**Target:** Binary (0 = No Disease, 1 = Disease)

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import joblib

# Define column names for the Cleveland dataset (since raw file has no headers)
column_names = [
    'age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 
    'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'num'
]

# Load the dataset
# Note: '?' in the dataset represents missing values, so we replace it with NaN
df = pd.read_csv('processed.cleveland.data', names=column_names, na_values='?')

print("Dataset loaded successfully!")
print(f"Total records: {df.shape[0]}")
```

```python
# Constraint: Select only 4 input attributes
selected_features = ['age', 'sex', 'cp', 'thalach']
X = df[selected_features].copy()

# Target transformation: Convert multi-class 'num' to binary
# 0 means no disease, 1,2,3,4 means presence of disease
y = (df['num'] > 0).astype(int)

# Handle missing values using the median strategy
imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X)

# Convert back to DataFrame to keep column names intact
X = pd.DataFrame(X_imputed, columns=selected_features)

print("Features selected:", selected_features)
print("Missing values after imputation:", X.isnull().sum().sum())
print("Target distribution:\n", y.value_counts())
```

```python
# Split the data into Training (80%) and Testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

```python
# Initialize and train the Logistic Regression model
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Save the trained model to a file for deployment
model_filename = 'heart_model.pkl'
joblib.dump(model, model_filename)

print(f"Model successfully saved as '{model_filename}'")
```

```python
# Test the model with a hypothetical patient
# Format: [[Age, Sex (1=Male, 0=Female), CP (1-4), Thalach]]
sample_patient = [[55, 1, 4, 140]] 

prediction = model.predict(sample_patient)

if prediction[0] == 1:
    print("Prediction: High Risk - Heart Disease Detected")
else:
    print("Prediction: Low Risk - No Heart Disease Detected")
```
